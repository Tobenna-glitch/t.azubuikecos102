```python
import tkinter as tk
from tkinter import messagebox

# 🔹 Parent Class
class Zenith:
    def mutual_services(self):
        return [
            "Lines of credit",
            "Investment management and accounts",
            "Insurance"
        ]

    def unique_services(self):
        return []


# 🔹 Child Classes (Inheritance + Polymorphism)
class RetailBanking(Zenith):
    def unique_services(self):
        return [
            "Retirement and education accounts",
            "Loans and mortgages",
            "Checking and saving"
        ]


class GlobalBanking(Zenith):
    def unique_services(self):
        return [
            "Multi-currency management services and products",
            "Foreign currency accounts",
            "Foreign currency credit cards",
            "Transborder advisory services",
            "Liquidity management"
        ]


class CommercialBanking(Zenith):
    def unique_services(self):
        return [
            "Advisory services"
        ]


# 🔹 Function to handle GUI input
def show_services():
    name = entry_name.get()
    division = entry_division.get().lower()

    # 🔹 Object creation (Polymorphism in action)
    if division == "retail":
        dept = RetailBanking()
    elif division == "global":
        dept = GlobalBanking()
    elif division == "commercial":
        dept = CommercialBanking()
    else:
        messagebox.showerror("Error", "Invalid division! Use Retail, Global or Commercial.")
        return

    # 🔹 Combine services
    services = dept.mutual_services() + dept.unique_services()

    result = f"Employee: {name}\nDivision: {division.capitalize()}\n\nServices:\n"
    for s in services:
        result += f"- {s}\n"

    messagebox.showinfo("Services", result)


# 🔹 GUI Setup
root = tk.Tk()
root.title("Zenith Bank Services")
root.geometry("400x300")

tk.Label(root, text="Employee Name").pack()
entry_name = tk.Entry(root)
entry_name.pack()

tk.Label(root, text="Division (Retail / Global / Commercial)").pack()
entry_division = tk.Entry(root)
entry_division.pack()

tk.Button(root, text="Show Services", command=show_services).pack(pady=10)

root.mainloop()
```
