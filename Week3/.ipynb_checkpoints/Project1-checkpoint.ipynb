```python



import tkinter as tk
from tkinter import messagebox
import csv

def check_employee():
    
    user_sur = entry1.get().lower().strip()
    user_first = entry2.get().lower().strip()
    user_dept = entry3.get().lower().strip()

    found = False

    
    try:
        with open('GIG-logistics.xlsx - in.csv', mode='r') as file:
            reader = csv.DictReader(file)
            
            for row in reader:
                # Check if the row matches the user input
                if row['SURNAME'].lower().strip() == user_sur and \
                   row['FIRST NAME'].lower().strip() == user_first and \
                   row['DEPARTMENT'].lower().strip() == user_dept:
                    
                    found = True
                    messagebox.showinfo("Success", "Welcome " + user_first)
                    break 

        if found == False:
            messagebox.showerror("Error", "Employee not found")
            
    except:
        messagebox.showerror("Error", "File not found!")


root = tk.Tk()
root.title("Verification")
root.geometry("600x400")


tk.Label(root, text="Surname:").pack()
entry1 = tk.Entry(root)
entry1.pack()

tk.Label(root, text="First Name:").pack()
entry2 = tk.Entry(root)
entry2.pack()

tk.Label(root, text="Department:").pack()
entry3 = tk.Entry(root)
entry3.pack()

btn = tk.Button(root, text="Check", command=check_employee)
btn.pack()

root.mainloop()
        

```
