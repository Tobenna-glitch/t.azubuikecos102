# Inheritance in Python

Inheritance is the capability of one class to derive or inherit the properties from another class. The benefits of inheritance are: 
 
<ol><li>
It represents real-world relationships well.</li><li>
It provides reusability of a code, so that you don’t have to write the same code again and again. </li><li>It allows for adding more features to a class without modifying it.</li><li>
It is transitive in nature, which means that if class B inherits from another class A, then all the subclasses of B would automatically inherit from class A.</li><ol>

### A Python program to demonstrate inheritance

```python
class Person(object):

    # Constructor
    def __init__(self, name):
        self.name = name  # Fixed: changed 'names' to 'name' to match the parameter

    # To get name method
    def getname(self):
        return self.name

    # To check if this person is an employee
    def isEmployee(self):
        return False
    

# Inherited or Subclass (Note Person in bracket)
class Employee(Person):

    # Here employee method returns true
    def isEmployee(self):
        return True

# Object Instantiation
emp = Person("Segun Da-Silver") # An Object of Person
print(emp.getname(), emp.isEmployee())

emp = Employee("Omotayo Ayeni") # An Object of Employee
print(emp.getname(), emp.isEmployee())
```

### Subclassing (Calling constructor of parent class) 
<ul><li>A child class needs to identify which class is its parent class.</li><li> This can be done by mentioning the parent class name in the definition of the child class. 
Eg: class subclass_name (superclass_name): </li></ul>

```python
# Python code to demonstrate how parent constructors
# are called.

# parent class
class Person(object):

        # __init__ is known as the constructor
        def __init__(self, name, idnumber):
            self.name = name
            self.idnumber = idnumber
            
        def display(self):
            print("Name:", self.name)
            print("ID No.:", self.idnumber)  # Fixed: removed comma and corrected variable name
            print("Salary:", self.salary)

# child class
class Employee(Person):  # Fixed: corrected class name from 'Persons' to 'Person'
    def __init__(self, name, idnumber, salary, post, age):
        self.salary = salary
        self.post = post

        # invoking the __init__ of the parent class
        Person.__init__(self, name, idnumber)  # Fixed: added missing arguments

                
# creation of an object variable or an instance
a = Employee('Adedoyin Adeniji', 886012, 200000, "Intern", 25)  # Fixed: added missing age argument

# calling a function of the class Person using its instance
a.display()
```

### Note that:
<ul><li>In Python, every class inherits from a built-in basic class called ‘object’. The constructor i.e. the ‘__init__’ function of a class is invoked when we create an object variable or an instance of the class.</li><li>
The variables defined within __init__() are called as the instance variables or objects. Hence, ‘name’ and ‘idnumber’ are the objects of the class Person. Similarly, ‘salary’ and ‘post’ are the objects of the class Employee.</li><li> Since the class Employee inherits from class Person, ‘name’ and ‘idnumber’ are also the objects of class Employee.</li><li>
If you forget to invoke the __init__() of the parent class then its instance variables would not be available to the child class. </li></ul>

```python
# Python program to demonstrate inheritance

class A:
    # class attribute
    surname = "Moru"

    def __init__(self, n='Emmanuel Abbah'):
        # instance attribute
        self.name = n


class B(A):
    def __init__(self, roll):
        super().__init__()   # Call parent constructor
        self.roll = roll


# Object Instance
project = B(23)

print(project.name)
print(project.roll)
print(project.surname)
```

## Types of Inheritance in Python
There are four types of inheritance in Python. The type of Inheritance depends upon the number of child and parent classes involved. : 

#### Single Inheritance:
<ul><li> Single inheritance enables a derived class to inherit properties from a single parent class, thus enabling code reusability and the addition of new features to existing code.</li></ul>

```python
# Python program to demonstrate
# single inheritance

# Base class
class Parent:
    def func1(self):
        print("This function is in parent class.")

# Derived class
class Child(Parent):
    def func2(self):
        print("This function is in child class.")

# Object Instance
obj = Child()

obj.func1()  # inherited from Parent
obj.func2()  # defined in Child
```

#### Multiple Inheritance:
<ul><li>When a class can be derived from more than one base class this type of inheritance is called multiple inheritance. </li><li>In multiple inheritance, all the features of the base classes are inherited into the derived class. </li></ul>
 

```python
# Python program to demonstrate
# multiple inheritance


# Base class1
class Mother: 
    mothername = "Caroline Aina"
    def mother(self):
        print(self.mothername)

# Base class2
class Father:
    fathername = "Paul Abiodun"
    def father(self):
        print(self.fathername) # Fixed typo: 'sef' to 'self'

# Derived class
class Son(Mother, Father):
    def parents(self):
        print("Father :", self.fathername)
        print("Mother :", self.mothername)

# Object Instance
s1 = Son() # Fixed case-sensitivity: 'son()' to 'Son()'
s1.parents()
```

#### Multilevel Inheritance
<ul><li>In multilevel inheritance, features of the base class and the derived class are further inherited into the new derived class.</li><li> This is similar to a relationship representing a child and grandfather.</li></ul> 

```python
# Python program to demonstrate
# multilevel inheritance

# Base class
class Grandfather:
    # Fixed typo: changed '_init__' to '__init__'
    def __init__(self, grandfathername):
        self.grandfathername = grandfathername

# Intermediate class
class Father(Grandfather):
    def __init__(self, fathername, grandfathername):
        self.fathername = fathername
        
        # invoking constructor of Grandfather class
        Grandfather.__init__(self, grandfathername)

# Derived class
class Son(Father):
    def __init__(self, sonname, fathername, grandfathername):
        self.sonname = sonname

        # invoking constructor of Father class
        Father.__init__(self, fathername, grandfathername)

    def print_name(self):
        print('Grandfather name :', self.grandfathername)
        print("Father name :", self.fathername)
        print("Son name :", self.sonname)

# Object Instance
s1 = Son('Gbenga', 'Oluwafemi', 'Adeoke')
print(s1.grandfathername)
s1.print_name()
```

#### Hierarchical Inheritance:
<ul><li>When more than one derived classes are created from a single base this type of inheritance is called hierarchical inheritance.</li><li> In this program, we have a parent (base) class and two child (derived) classes.</li></ul>

```python
# Python program to demonstrate
# Hierarchical inheritance

# Base class
class Parent:
    def func1(self):
        print("This function is in parent class.")

# Derived class1
class Child1(Parent): # Fixed: 'Parents' to 'Parent'
    def func2(self):
        print("This function is in child 1.")

# Derived class2
class Child2(Parent): # Fixed: 'Parents' to 'Parent'
    def func3(self):
        print("This function is in child 2.")

# Object Instantiation
object1 = Child1()
object2 = Child2() # Fixed: 'Child3()' to 'Child2()'

object1.func1() # Fixed: removed trailing underscore from object1_
object1.func2() # Fixed: removed trailing underscore from func2_

object2.func1() # Fixed: removed trailing underscore from object2_
object2.func3() # Fixed: removed trailing underscore from func3_
```

#### Hybrid Inheritance:
<ul><li>Inheritance consisting of multiple types of inheritance is called hybrid inheritance.</li></ul>

```python
# Python program to demonstrate
# hybrid inheritance

class School:
    def func1(self): # Fixed: changed 'Def' to 'def'
        print("This function is in school.")

class Student1(School):
    def func2(self): # Fixed: changed 'Def' to 'def'
        print("This function is in student 1.")

class Student2(School):
    def func3(self): # Fixed: changed 'Def' to 'def'
        print("This function is in student 2.")

# Student3 inherits from both Student1 and Student2 (which both inherit from School)
class Student3(Student1, Student2): 
    def func4(self): # Fixed: changed 'Def' to 'def'
        print("This function is in student 3.")

# Object Instantiation
obj = Student3() # Fixed: removed 'object' argument and changed variable name to 'obj'
obj.func1()
obj.func2()
obj.func3()
obj.func4()
```

# Polymorphism in Python

<ul><li>In programming, Polymorphism is a concept of OOP.</li><li>The word polymorphism means having many forms.</li><li>
    It enables using a single interface with the input of different data types, different classes or maybe for a different number of inputs.</li><li>Polymorphism means the same function name (but different signatures) being used for different types.</li></ul>

### Inbuilt polymorphic functions

```python
# Python program to demonstrate in-built polymorphic functions

# len() being used for a string
print(len("Department of Computer Science"))

# len() being used for a list
print(len([10, 20, 30, 40, 50, 60, 70]))
```

### User-defined polymorphic functions : 

```python
# A simple Python function to demonstrate
# Polymorphism (Method Overloading via default arguments)

def add(x, y, z = 0, w = 0): # Fixed: Given z and w default values of 0
    return x + y + z + w     # Fixed: Included all parameters in the return statement

# Driver code
print(add(2, 3))         # Uses default z=0, w=0 -> 2 + 3 + 0 + 0
print(add(2, 3, 4))      # Uses default w=0      -> 2 + 3 + 4 + 0
print(add(2, 3, 5, 6))   # Uses all passed values -> 2 + 3 + 5 + 6
```

### Polymorphism with class methods: 
<ul><li>The code below shows how Python can use two different class types, in the same way.</li><li> We create a for loop that iterates through a tuple of objects. Then call the methods without being concerned about which class type each object is. We assume that these methods actually exist in each class.</li></ul> 

```python
class Nigeria():
    def capital(self):
        print("Abuja is the capital of Nigeria.") # Updated for factual accuracy

    def language(self):
        print("English is the official language of Nigeria.")

    def type(self):
        print("Nigeria is the giant of Africa.")

class Togo():
    def capital(self):
        print("Lome is the capital of Togo.")

    def language(self):
        print("French is the primary language of Togo.")

    def type(self):
        print("Togo is known for its palm-lined beaches and hilltop villages.")

obj_naija = Nigeria()
obj_togo = Togo()

# Fixed: changed 'obj_naij' to 'obj_naija'
for country in (obj_naija, obj_togo):
    country.capital()   # Fixed: lowercase 'c'
    country.language()  # Fixed: removed the 's' at the end
    country.type()      # Fixed: lowercase 't'
    print("-" * 40)     # Added for clean separation in the output
```

### Polymorphism with Inheritance:
<ul><li>In Python, Polymorphism lets us define methods in the child class that have the same name as the methods in the parent class. In inheritance, the child class inherits the methods from the parent class.</li><li> However, it is possible to modify a method in a child class that it has inherited from the parent class.</li><li> This is particularly useful in cases where the method inherited from the parent class doesn’t quite fit the child class. In such cases, we re-implement the method in the child class.</li><li> This process of re-implementing a method in the child class is known as <b>Method Overriding. </b> </li></ul>

```python
class Bird:
    def intro(self):
        print("There are many types of birds.")

    def flight(self):
        print("Most of the birds can fly but some cannot.")

class Sparrow(Bird): # Fixed: Capitalized 'Sparrow'
    def flight(self):
        print("Sparrows can fly.")

class Ostrich(Bird): # Fixed: Capitalized 'Ostrich'
    def flight(self): # Fixed: Changed 'selfie' to 'self'
        print("Ostriches cannot fly.")

# Object Instantiation
obj_bird = Bird()    # Fixed: 'Birds()' to 'Bird()'
obj_spr = Sparrow()  # Fixed: Match capitalized class name
obj_ost = Ostrich()  # Fixed: Match capitalized class name

# Calling methods
obj_bird.intro()
obj_bird.flight()    # Fixed: 'flihgt()' typo

obj_spr.intro()
obj_spr.flight()     # Fixed: removed trailing 's' from 'flights()'

obj_ost.intro()
obj_ost.flight()     # Fixed: changed 'Flight()' to lowercase 'flight()'
```

```python
import tkinter as tk
from tkinter import ttk, messagebox

# --- OOP ARCHITECTURE ---

class Zenith:
    """Parent Class representing the bank's core identity."""
    def __init__(self, employee_name, division_name):
        self.employee_name = employee_name
        self.division_name = division_name

    def mutual_services(self):
        """Base method for shared services across common divisions."""
        return [
            "Lines of credit",
            "Investment management and accounts",
            "Insurance"
        ]

    def unique_services(self):
        """Placeholder method to be overridden by subclasses (Polymorphism)."""
        return []


class RetailBanking(Zenith):
    """Child Class 1 - Inherits from Zenith"""
    def unique_services(self):
        # Overriding the parent method with Retail-specific services
        return [
            "Retirement and education accounts",
            "Loans and mortgages",
            "Checking and savings"
        ]


class GlobalBanking(Zenith):
    """Child Class 2 - Inherits from Zenith"""
    def mutual_services(self):
        # Global banking doesn't share the same mutual services as Retail/Commercial
        return ["Liquidity management"]

    def unique_services(self):
        # Overriding the parent method with Global-specific services
        return [
            "Multi-currency management services and products",
            "Foreign currency accounts",
            "Foreign currency credit cards",
            "Transborder advisory services"
        ]


class CommercialBanking(Zenith):
    """Child Class 3 - Inherits from Zenith"""
    def unique_services(self):
        # Overriding the parent method with Commercial-specific services
        return ["Advisory services"]


# --- GUI IMPLEMENTATION ---

class ZenithGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("Zenith Bank Nigeria - Division Services Portal")
        self.root.geometry("550x500")
        self.root.resizable(False, False)
        
        # Style configurations
        self.style = ttk.Style()
        self.style.configure("TLabel", font=("Arial", 11))
        self.style.configure("TButton", font=("Arial", 11, "bold"))
        
        self.create_widgets()

    def create_widgets(self):
        # Title Header
        header = tk.Label(self.root, text="ZENITH BANK PLC", font=("Arial", 16, "bold"), fg="#E60000")
        header.pack(pady=10)
        
        subtitle = tk.Label(self.root, text="Employee & Division Services Directory", font=("Arial", 10, "italic"))
        subtitle.pack(pady=2)

        # Input Frame
        input_frame = ttk.LabelFrame(self.root, text=" Verification Portal ", padding=15)
        input_frame.pack(pady=15, padx=20, fill="x")

        # Employee Name Input
        ttk.Label(input_frame, text="Employee Name:").grid(row=0, column=0, sticky="w", pady=5)
        self.entry_name = ttk.Entry(input_frame, width=35)
        self.entry_name.grid(row=0, column=1, pady=5, padx=10)

        # Division Dropdown Selection
        ttk.Label(input_frame, text="Select Division:").grid(row=1, column=0, sticky="w", pady=5)
        self.combo_division = ttk.Combobox(input_frame, width=32, state="readonly")
        self.combo_division['values'] = ("Retail Banking", "Global Banking", "Commercial Banking")
        self.combo_division.grid(row=1, column=1, pady=5, padx=10)
        self.combo_division.current(0)

        # Submit Button
        self.btn_submit = ttk.Button(input_frame, text="Fetch Services", command=self.process_request)
        self.btn_submit.grid(row=2, column=0, columnspan=2, pady=12)

        # Output Display Frame
        self.output_frame = ttk.LabelFrame(self.root, text=" Division Profile & Rendered Services ", padding=15)
        self.output_frame.pack(pady=5, padx=20, fill="both", expand=True)

        self.txt_display = tk.Text(self.output_frame, wrap="word", font=("Consolas", 10), bg="#F9F9F9", state="disabled")
        self.txt_display.pack(fill="both", expand=True)

    def process_request(self):
        emp_name = self.entry_name.get().strip()
        division = self.combo_division.get()

        if not emp_name:
            messagebox.showerror("Input Error", "Please enter the employee's name.")
            return

        # POLYMORPHISM IN ACTION: 
        # Instantiating the correct subclass dynamically based on user choice
        if division == "Retail Banking":
            division_obj = RetailBanking(emp_name, division)
        elif division == "Global Banking":
            division_obj = GlobalBanking(emp_name, division)
        else:
            division_obj = CommercialBanking(emp_name, division)

        self.display_results(division_obj)

    def display_results(self, division_obj):
        """Interacts polymorphically with whatever object subclass is passed in."""
        self.txt_display.config(state="normal")
        self.txt_display.delete("1.0", tk.END)

        # Header Information
        profile_info = (
            f"Employee Name:     {division_obj.employee_name}\n"
            f"Assigned Division: {division_obj.division_name}\n"
            f"{'='*45}\n\n"
            f"SERVICES RENDERED:\n\n"
        )
        self.txt_display.insert(tk.END, profile_info)

        # Polymorphic compilation of services
        self.txt_display.insert(tk.END, "--- Mutual/Core Services ---\n")
        for service in division_obj.mutual_services():
            self.txt_display.insert(tk.END, f" * {service}\n")
            
        self.txt_display.insert(tk.END, "\n--- Unique/Exclusive Services ---\n")
        for service in division_obj.unique_services():
            self.txt_display.insert(tk.END, f" * {service}\n")

        self.txt_display.config(state="disabled")


# --- RUN APPLICATION ---
if __name__ == "__main__":
    root = tk.Tk()
    app = ZenithGUI(root)
    root.mainloop()
```
