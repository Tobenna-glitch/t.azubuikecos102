```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


```

```python
df = pd.read_csv("C://Users//Rora//Downloads//test.csv")


print("First 5 rows of the dataset:")
print(df.head())


print("\nDataset Info:")
print(df.info())


print("\nMissing Values Count:")
print(df.isnull().sum())
```

```python

df['Age'] = df['Age'].fillna(df['Age'].median())


df['Fare'] = df['Fare'].fillna(df['Fare'].median())

df_cleaned = df.drop(columns=['Cabin'])

```

```python

avg_fare = df_cleaned['Fare'].mean()
print(f"\nAverage Fare: ${avg_fare:.2f}")


pclass_counts = df_cleaned['Pclass'].value_counts().sort_index()
print("\nPassenger Count by Class:")
print(pclass_counts)
```

```python
# 1. Histogram of Passenger Ages
plt.figure(figsize=(10, 6))
plt.hist(df_cleaned['Age'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')

```

```python
# 2. Bar Chart of Passenger Classes
plt.figure(figsize=(10, 6))
colors = ['#FFD700', '#C0C0C0', '#CD7F32'] # Gold, Silver, Bronze for classes
pclass_counts.plot(kind='bar', color=colors, edgecolor='black')
plt.title('Number of Passengers in Each Class')
plt.xlabel('Passenger Class ($1^{st}$, $2^{nd}$, $3^{rd}$)')
plt.ylabel('Count')
plt.xticks(rotation=0)


```
