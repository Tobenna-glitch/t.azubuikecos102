```python
import tkinter as tk
from tkinter import messagebox


# Delivery Class
class DeliveryService:
    def __init__(self, location, weight):
        self.location = location
        self.weight = weight

    def calculate_charge(self):
        if self.location.upper() == "PAU":
            if self.weight >= 10:
                return 2000
            else:
                return 1500

        elif self.location.upper() == "EPE":
            if self.weight >= 10:
                return 5000
            else:
                return 4000

        else:
            return None


# Function for GUI
def get_charge():
    try:
        location = location_entry.get().strip()
        weight = float(weight_entry.get())

        delivery = DeliveryService(location, weight)
        charge = delivery.calculate_charge()

        if charge is not None:
            result_label.config(
                text=f"Delivery Charge: ₦{charge:,}",
                fg="green"
            )
        else:
            result_label.config(
                text="Sorry! We only deliver to PAU and EPE.",
                fg="red"
            )

    except ValueError:
        messagebox.showerror(
            "Input Error",
            "Please enter a valid package weight."
        )


# Main Window
root = tk.Tk()
root.title("Delivery Charge Calculator")
root.geometry("400x300")

# Heading
heading = tk.Label(
    root,
    text="Delivery Service Charge Calculator",
    font=("Arial", 14, "bold")
)
heading.pack(pady=10)

# Location Input
tk.Label(root, text="Enter Location (PAU or EPE):").pack()
location_entry = tk.Entry(root, width=30)
location_entry.pack(pady=5)

# Weight Input
tk.Label(root, text="Enter Package Weight (kg):").pack()
weight_entry = tk.Entry(root, width=30)
weight_entry.pack(pady=5)

# Button
calculate_btn = tk.Button(
    root,
    text="Calculate Charge",
    command=get_charge
)
calculate_btn.pack(pady=15)

# Result Label
result_label = tk.Label(root, text="", font=("Arial", 12))
result_label.pack(pady=10)

# Run Application
root.mainloop()
```
