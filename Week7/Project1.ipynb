```python
import tkinter as tk
from tkinter import messagebox
import random

# Employee Class
class Employee:
    employees = [
        "Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali",
        "Andrew Ugwu", "Stella Mankinde", "Jane Akibo", "Ago James",
        "Michell Taiwo", "Abraham Jones", "Nicole Anide",
        "Kosi Korso", "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"
    ]

    tasks = [
        "Loading",
        "Transporting",
        "Reviewing Orders",
        "Customer Service",
        "Delivering Items"
    ]

    attendance = []

    def __init__(self, name):
        self.name = name

    def check_employee(self):
        return self.name in Employee.employees

    def take_attendance(self):
        if self.name not in Employee.attendance:
            Employee.attendance.append(self.name)
        return f"Attendance taken for {self.name}"

    def assign_task(self):
        task = random.choice(Employee.tasks)
        return task

    def refuse_access(self):
        return f"Access Denied! {self.name} is not an employee."


# GUI Functions
def verify_employee():
    name = entry_name.get().strip()

    if not name:
        messagebox.showwarning("Input Error", "Please enter a name.")
        return

    employee = Employee(name)

    if employee.check_employee():
        attendance_msg = employee.take_attendance()
        task = employee.assign_task()

        result = (
            f"Employee Verified!\n\n"
            f"Name: {name}\n"
            f"{attendance_msg}\n"
            f"Assigned Task: {task}"
        )

        label_result.config(text=result)
    else:
        label_result.config(text=employee.refuse_access())


# Main Window
root = tk.Tk()
root.title("Employee Attendance and Task Assignment System")
root.geometry("500x350")

# Heading
heading = tk.Label(
    root,
    text="Employee Verification System",
    font=("Arial", 16, "bold")
)
heading.pack(pady=10)

# Name Entry
label_name = tk.Label(root, text="Enter Employee Name:")
label_name.pack()

entry_name = tk.Entry(root, width=40)
entry_name.pack(pady=5)

# Verify Button
btn_verify = tk.Button(
    root,
    text="Verify Employee",
    command=verify_employee,
    bg="lightblue"
)
btn_verify.pack(pady=10)

# Result Display
label_result = tk.Label(
    root,
    text="",
    font=("Arial", 11),
    justify="left"
)
label_result.pack(pady=20)

# Run GUI
root.mainloop()
```
